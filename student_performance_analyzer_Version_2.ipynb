```python
students = [
    {"name": "Gowtham", "maths": 95, "python": 88, "english": 95},
    {"name": "Rahul", "maths": 72, "python": 80, "english": 75},
    {"name": "Priya", "maths": 90, "python": 94, "english": 88}
]
```

```python
students = []

student = {
    "name":"gowtham",
    "maths":95,
    "english":88,
    "python":95
}
student_2 ={
    "name":"rahul",
    "maths":72,
    "english":75,
    "python":80
}
student_3 = {
    "name":"priya",
    "maths":90,
    "english":88,
    "python":94
}
students.append(student)
students.append(student_2)
students.append(student_3)
print(students)
```

```python
for student in students:
  print(student["name"])
```

```python
for student in students:
    total = student["maths"] + student["python"] + student["english"]
    print(student["name"], "Total:", total)
```

```python
for student in students:
    total = student["maths"] + student["python"] + student["english"]
    average = total / 3
    percentage = total / 300 * 100

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    if percentage >= 50:
        result = "Pass"
    else:
        result = "Fail"

    print(student["name"])
    print("Total:", total)
    print("Average:", round(average, 2))
    print("Percentage:", round(percentage, 2))
    print("Grade:", grade)
    print("Result:", result)
    print()
```
