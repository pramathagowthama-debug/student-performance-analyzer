```python
name = input("Enter student name: ")
maths = int(input("Enter Maths marks: "))
python = int(input("Enter Python marks: "))
english = int(input("Enter English marks: "))
```

```python
total = maths + python + english
average = total / 3
percentage = total / 300 * 100

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

if percentage >= 50:
    result = "Pass"
else:
    result = "Fail"
```

```python
print("\n===== STUDENT REPORT =====")
print("Name:", name)
print("Maths:", maths)
print("Python:", python)
print("English:", english)
print("Total:", total)
print("Average:", round(average,2))
print("Percentage:", round(percentage,2))
print("Grade:", grade)
print("Result:", result)
```
